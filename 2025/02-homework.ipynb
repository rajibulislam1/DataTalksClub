```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('car_fuel_efficiency.csv')
df.head()
```

```python
df_filtered = df[['engine_displacement','horsepower','vehicle_weight','model_year','fuel_efficiency_mpg']]
df_filtered.head()
```

### EDA

```python
plt.figure(figsize=(10, 6))
sns.histplot(df_filtered['fuel_efficiency_mpg'], edgecolor = 'black', alpha = 0.7, bins=30)
plt.ylabel('Count')
plt.xlabel('Fuel Efficiency (MPG)')
plt.title('Distribution of Fuel Efficiency (MPG)')
plt.show()
```

```python
df_filtered.boxplot(column='fuel_efficiency_mpg')
plt.title('Boxplot of Fuel Efficiency')
plt.show()
```

```python
df_filtered['fuel_efficiency_mpg'].plot(kind='kde')
plt.title('Density Plot of Fuel Efficiency')
plt.show()
```

```python
df_filtered['fuel_efficiency_mpg'].skew()
```

```python
df_filtered['fuel_efficiency_mpg'].kurt()
```

```python
df['fuel_efficiency_mpg'].mean(), df['fuel_efficiency_mpg'].median()

# If mean > median → right-skewed (long right tail)

# If mean < median → left-skewed (long left tail)
```
