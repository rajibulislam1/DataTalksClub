```python
# 1. Median value of horsepower column
df['horsepower'].median()
```

```python
# 2. the most frequent value of the same horsepower column.
most_frequent = df['horsepower'].mode()[0]
most_frequent
```

```python
# 3.0. check how many missing values are there in the horsepower column
df['horsepower'].isnull().sum()
```

```python
# 3.0 fill the missing values in the horsepower column with the most frequent value
df['horsepower'] = df['horsepower'].fillna(most_frequent)
df['horsepower']
```

```python
# 4.0 Calculate median value of horsepower again
df['horsepower'].median()
```
